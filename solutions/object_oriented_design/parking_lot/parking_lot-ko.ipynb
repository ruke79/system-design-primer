## 제약 조건과 가정

* 어떤 종류의 차량을 지원해야 하나요?
    * 오토바이(Motorcycle), 자동차(Car), 버스(Bus)
* 차량 종류마다 차지하는 주차 공간의 수가 다른가요?
    * 네
    * 오토바이 자리(Motorcycle spot) -> 오토바이
    * 소형 자리(Compact spot) -> 오토바이, 자동차
    * 대형 자리(Large spot) -> 오토바이, 자동차
    * 버스는 연속된 "대형" 자리 5개가 있어야 주차 가능
* 주차장은 여러 층으로 되어 있나요?
    * 네
